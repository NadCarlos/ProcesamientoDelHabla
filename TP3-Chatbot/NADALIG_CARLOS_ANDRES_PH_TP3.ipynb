###  2) Crear el chatbot utilizando TFIDF y similitud del coseno. (1 punto)

Todo el codigo fue recuperado desde los GoogleColab facilitado por la docente y luego refactorizados.

```python
stopwords_es = [
    "a", "al", "como", "con", "cual", "de", "del", "donde",
    "el", "en", "es", "esta", "hay", "la", "las", "lo", "los",
    "para", "por", "puedo", "que", "se", "si", "tiene", "una", "un", "y", "con", "que",
    "los", "para", "un", "una", "el", "la", "en", "y", "o", "de", "a", "se", "es", "al",
    "como", "por", "no", "su", "más", "pero", "si", "este", "esta", "eso", "esa", "todo",
    "todos", "todas", "cada", "cual", "cualquier", "algo", "alguna", "algún", "ningún", 
    "ninguna", "ninguno", "otro", "otros", "otras", "sobre", "entre", "durante", "desde", 
    "hasta", "también", "así", "mismo", "misma", "mismo", "misma", "tan", "tanto", "tanta", 
    "muy", "poco", "poca", "poco", "nada", "solo", "sola", "solo", "sola", "siempre", "nunca", 
    "jamás", "casi", "aproximadamente", "cerca", "lejos", "antes", "después", "ahora", "hoy", 
    "ayer", "mañana", "luego", "mientras", "mientras tanto", "después de", "antes de", "durante", 
    "sin", "aunque",     "a pesar de", "debido a", "gracias a", "por qué", "cómo", "dónde",
    "cuándo", "quién", "qué", "cuál",
]
vectorizador_tfidf = TfidfVectorizer(
    lowercase=True,
    strip_accents="unicode",
    stop_words=stopwords_es,
    ngram_range=(1, 2),
)
matriz_tfidf = vectorizador_tfidf.fit_transform(preguntas)
```

```python
matriz_tfidf.shape
```

```python
def chatbot_tfidf(consulta_usuario, umbral=0.3, mostrar_detalle=True):
    consulta_vector = vectorizador_tfidf.transform([consulta_usuario])
    similitudes = cosine_similarity(consulta_vector, matriz_tfidf)[0]
    indice = int(np.argmax(similitudes))
    mejor_similitud = float(similitudes[indice])

    if mostrar_detalle == True:
        if mejor_similitud < umbral:
            print("Similitud:", round(mejor_similitud, 3))
            return print("Respuesta: No encontré una respuesta suficientemente relacionada. ¿Podés reformular la consulta?")
        print("Pregunta recuperada:", preguntas[indice])
        print("Similitud:", round(mejor_similitud, 3))
        return print("Respuesta:", respuestas[indice])
        

    if mostrar_detalle == False:
        if mejor_similitud < umbral:
            return respuestas[indice], preguntas[indice], "No encontré una respuesta suficientemente relacionada. ¿Podés reformular la consulta?"
        return respuestas[indice], preguntas[indice], round(mejor_similitud, 3)
```

### 3) Chatbot con embeddings preentrenados

Para el segundo chatbot usé el modelo "es_core_news_md" de spaCy. Elegi este modelo porque está entrenado para trabajar con textos en español e incluye vectores de palabras preentrenados permitiendo representar las consultas mediante vectores y comparar su similitud semántica.

Armamos vector semántico preentrenado de cada frase.

```python
def obtener_embedding(texto):
    documento = nlp(texto)
    if documento.vector_norm == 0:
        return np.zeros(nlp.vocab.vectors_length, dtype=np.float32)
    return documento.vector
```

```python
embeddings_preguntas = np.vstack([
    obtener_embedding(pregunta) for pregunta in preguntas
])
```

```python
def chatbot_embeddings(consulta_usuario, umbral=0.7, mostrar_detalle=True):
    embedding_usuario = obtener_embedding(consulta_usuario)
    if not np.any(embedding_usuario):
        return "No pude representar esa consulta. ¿Podés reformularla?"

    similitudes = cosine_similarity(embedding_usuario.reshape(1, -1), embeddings_preguntas)[0]
    indice = int(np.argmax(similitudes))
    mejor_similitud = float(similitudes[indice])

    if mostrar_detalle == True:
        if mejor_similitud < umbral:
            print("Similitud:", round(mejor_similitud, 3))
            return print("Respuesta: No encontré una respuesta suficientemente relacionada. ¿Podés reformular la consulta?")
        print("Pregunta recuperada:", preguntas[indice])
        print("Similitud:", round(mejor_similitud, 3))
        return print("Respuesta:", respuestas[indice])

    if mostrar_detalle == False:
        if mejor_similitud < umbral:
            return respuestas[indice], preguntas[indice], "No encontré una respuesta suficientemente relacionada. ¿Podés reformular la consulta?"
        return respuestas[indice], preguntas[indice], round(mejor_similitud, 3)
```

### Errores y problemas encontrados

**Chatbot TF-IDF**

Aunque obtuvo mejores resultados, una de las principales limitaciones de TF-IDF es que depende principalmente de la coincidencia de palabras entre la consulta y las preguntas almacenadas, al usar alguna expresion diferente se va a equivocar.
Algunos errores:

- "¿Dónde queda el local?": devolvió una respuesta relacionada con los horarios de atención.
- "¿Qué motos tienen disponibles?": confundió la disponibilidad de modelos con información sobre la garantía.
- "¿Tienen motos para entregar ya?": nuevamente recuperó información sobre la garantía.
- "Necesito hacerle el service": devolvió información sobre los requisitos de financiación.
- "Quiero pagar una parte y financiar lo demás": devolvio una respuesta relacionada al pago, pero no al financiamiento.

**Chatbot con embeddings**

El chatbot basado en embeddings presentó una mayor cantidad de errores, incluso en consultas donde la intención parecía relativamente clara.
varias respuestas incorrectas obtuvieron valores de similitud elevados, incluso superiores a 0.90 demuestrando que una similitud vectorial alta no garantiza que la respuesta recuperada sea correcta.
Algunos errores:

- "¿A qué hora abren?": respondió sobre la disponibilidad de entrega inmediata.
- "Quiero una moto para viajar": recuperó información sobre la documentación necesaria para comprar una moto.
